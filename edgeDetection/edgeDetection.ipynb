```python
#import
import cv2
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image, ImageChops
```

**Lab Task 1**
Starting from any grayscale image of your choice, compute the Laplacian, then create a second
version by first applying a Gaussian blur (kernel size 5×5, sigma = 1.5) before the Laplacian.
Display the original Laplacian and the blurred-then-Laplacian results next to each other.

**Observation:**
The laplacian before applying gaussian filter was giving an image with edges not as well defined.
After gaussian filter is applied the image edges are more visible when computing the laplacian.

```python
image_path = "cat.jpg"
rgb_image = Image.open(image_path).convert("RGB")
gray_image = rgb_image.convert("L")
rgb_array = np.array(rgb_image)
gray_array = np.array(gray_image)
print("RGB shape:", rgb_array.shape)
print("Grayscale shape:", gray_array.shape)
# Display RGB and grayscale images
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(rgb_array)
axes[0].set_title("RGB Image")
axes[0].axis("off")
axes[1].imshow(gray_array, cmap="gray")
axes[1].set_title("Grayscale Image")
axes[1].axis("off")
plt.tight_layout()
plt.show()
```

```python
#computing the laplacian without gaussian
#applying laplacian on image using cv2:
# Load image in grayscale
image = cv2.imread('cat.jpg', cv2.IMREAD_GRAYSCALE)

# Compute Laplacian using a higher precision data type to avoid clipping negative values
laplacian = cv2.Laplacian(image, cv2.CV_16S, ksize=3)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(image, cmap="gray")
axes[0].set_title("GrayScale Image")
axes[0].axis("off")
axes[1].imshow(laplacian, cmap="gray")
axes[1].set_title("laplacian Image")
axes[1].axis("off")
plt.tight_layout()
plt.show()
```

```python
#applying gaussian blur to image
#applying gaussian filter using cv2:
blurred = cv2.GaussianBlur(image, (5, 5), 1.5,)

laplacian_signed = cv2.Laplacian(blurred, cv2.CV_16S, ksize=3)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(blurred, cmap="gray")
axes[0].set_title("Gaussian Image")
axes[0].axis("off")
axes[1].imshow(laplacian_signed, cmap="gray")
axes[1].set_title("laplacian after Gaussian Image")
axes[1].axis("off")
plt.tight_layout()
plt.show()
```

```python
#comparison of laplacian images:
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(laplacian, cmap="gray")
axes[0].set_title("Laplacian Image")
axes[0].axis("off")
axes[1].imshow(laplacian_signed, cmap="gray")
axes[1].set_title("laplacian after Gaussian Image")
axes[1].axis("off")
plt.tight_layout()
plt.show()
```

**Lab Task 2**
Run the Canny edge detector on any image of your choice, three separate times using the
threshold pairs (50, 100), (100, 200) and (150, 300). Arrange the three binary edge maps in a
single figure with three sub-plots, label each sub-plot with its threshold values,

```python
#running canny edge detection on an image with three seperate thresholds:
#Edge Detection
edges_1 = cv2.Canny(image, 50, 100)
plt.imshow(edges_1, cmap='gray')
plt.title("Edge Map for values 50 and 100")
plt.axis('off')
plt.show()
```

```python
#Edge Detection
edges_2 = cv2.Canny(image, 100, 200)
plt.imshow(edges_2, cmap='gray')
plt.title("Edge Map for values 100 and 200")
plt.axis('off')
plt.show()
```

```python
#Edge Detection
edges_3 = cv2.Canny(image, 150, 300)
plt.imshow(edges_3, cmap='gray')
plt.title("Edge Map for values 150 and 300")
plt.axis('off')
plt.show()
```

```python
#mapped side by side:
fig, axes = plt.subplots(1, 3, figsize=(12, 5))
axes[0].imshow(edges_1, cmap="gray")
axes[0].set_title("Edge map for values 50 and 100")
axes[0].axis("off")
axes[1].imshow(edges_2, cmap="gray")
axes[1].set_title("Edge map for values 100 and 200")
axes[1].axis("off")
axes[2].imshow(edges_3, cmap='gray')
axes[2].set_title("Edge Map for values 150 and 300")
axes[2].axis('off')
plt.tight_layout()
plt.show()
```
